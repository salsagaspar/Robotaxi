# Data Understanding

```python
import pandas as pd
import numpy as np

# 1. Load Data
print("Loading data...")
rides = pd.read_csv('robotaxi_rides.csv')
users = pd.read_csv('robotaxi_users.csv')

# 2. Data Understanding
print('--- DATA UNDERSTANDING ---')
print(f'Rides shape: {rides.shape}')
print(f'Users shape: {users.shape}\n')

# Menampilkan 10 baris pertama data perjalanan (rides)
print("Preview Data Rides:")
display(rides.head(10))

# Menampilkan 10 baris pertama data pengguna (users)
print("\nPreview Data Users:")
display(users.head(10))
```

# Data Cleaning & Preparation
mengecek apakah ada data yang kosong (missing values), mengatasinya, lalu menggabungkan (join) kedua tabel tersebut menjadi satu dataframe utuh.

```python
# --- 3. DATA CLEANING & PREPARATION ---

print('--- CEK MISSING VALUES ---')
print("Missing values di data Rides:")
print(rides.isnull().sum()[rides.isnull().sum() > 0])

print("\nMissing values di data Users:")
print(users.isnull().sum()[users.isnull().sum() > 0])

# Mengatasi data kosong: 
# Kita isi user_rating yang kosong (NaN) dengan angka -1 agar tidak error saat dihitung
rides['user_rating'] = rides['user_rating'].fillna(-1)

# Menggabungkan data (Join) berdasarkan 'user_id'
df = pd.merge(rides, users, on='user_id', how='left')

print('\n--- PENGGABUNGAN DATA (JOIN) ---')
print(f'Ukuran data setelah digabung: {df.shape}')

# Cek hasil akhir tabel yang sudah digabung
display(df.head(10))
```

# Exploratory Data Analysis (EDA)

**Fokus 1: Operasional (Status Perjalanan & Waktu Tunggu)**

```python
# --- 4. EXPLORATORY DATA ANALYSIS (EDA) ---
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print('=== EDA FOKUS 1: OPERASIONAL ===\n')

# 1. Menghitung persentase tiap status perjalanan
status_counts = df['ride_status'].value_counts(normalize=True) * 100
print('Distribusi Status Perjalanan (%):')
display(status_counts.round(2).to_frame(name='Persentase (%)'))

# 2. Melihat rata-rata waktu tunggu berdasarkan status perjalanan
print('\nRata-rata Waktu Tunggu (Menit) berdasarkan Status:')
wait_time_status = df.groupby('ride_status')['wait_time_minutes'].mean().round(2)
display(wait_time_status.to_frame(name='Avg Wait Time'))

# 3. Visualisasi sederhana dengan Bar Chart
plt.figure(figsize=(10, 5))
sns.barplot(x=wait_time_status.index, y=wait_time_status.values, palette='viridis')
plt.title('Rata-rata Waktu Tunggu berdasarkan Status Perjalanan')
plt.ylabel('Menit')
plt.xlabel('Status')
plt.show()
```

**Insight untuk Fokus 1: Operasional (Status & Waktu Tunggu)**
1. Tingkat Kesuksesan (Completion Rate) yang Sehat: Robotaxi berhasil menyelesaikan hampir 77% dari total pesanan. Namun, angka Cancelled (12,1%) dan No-show (5,2%) jika digabung menjadi ~17,3%. Ini adalah potensi pendapatan yang hilang yang harus dievaluasi.

2. Waktu Tunggu Bukan Penyebab Utama Pembatalan: Rata-rata waktu tunggu untuk pesanan yang "Selesai" (5,98 menit) ternyata sama dengan pesanan yang "Dibatalkan" (5,97 menit).

3. Rekomendasi Bisnis: Karena pelanggan tidak membatalkan karena mobil terlalu lama datang, kemungkinan pembatalan terjadi karena:
- Perubahan pikiran (change of mind) dari penumpang.
- Kesalahan titik jemput (pickup location).

Saran: Terapkan biaya pembatalan (cancellation fee) jika pesanan dibatalkan setelah 2-3 menit mobil dipesan untuk mengurangi pembatalan impulsif.

**Fokus 2: Marketing & Pendapatan (Segmentasi User)**

```python
print('=== EDA FOKUS 2: MARKETING & ARPU ===\n')

# 1. Menghitung Total Pendapatan (Tarif + Tip)
df['total_revenue'] = df['fare_amount_usd'] + df['tip_amount_usd']

# 2. Menghitung Persentase Tip dari Tarif
df['tip_percentage'] = np.where(
    df['fare_amount_usd'] > 0, 
    (df['tip_amount_usd'] / df['fare_amount_usd']) * 100, 
    0
)

# KITA HANYA MENGANALISIS PERJALANAN YANG 'COMPLETED' (Selesai)
completed_rides = df[df['ride_status'] == 'Completed']

# 3. Rata-rata Pendapatan per Tier
print('Rata-rata Pendapatan ($) berdasarkan Loyalty Tier:')
arpu_tier = completed_rides.groupby('loyalty_tier')['total_revenue'].mean().round(2).sort_values(ascending=False)
display(arpu_tier.to_frame(name='Avg Revenue ($)'))

# 4. Rata-rata Persentase Tip per Tier
print('\nRata-rata Persentase Tip (%) berdasarkan Loyalty Tier:')
tip_tier = completed_rides.groupby('loyalty_tier')['tip_percentage'].mean().round(2).sort_values(ascending=False)
display(tip_tier.to_frame(name='Avg Tip (%)'))

# 5. Visualisasi Pendapatan vs Tier
plt.figure(figsize=(10, 5))
# Kita urutkan dari tier terendah ke tertinggi
order_tier = ['Bronze', 'Silver', 'Gold', 'Platinum', 'Diamond']
sns.barplot(data=completed_rides, x='loyalty_tier', y='total_revenue', order=order_tier, palette='magma', errorbar=None)
plt.title('Rata-rata Pendapatan (Tarif+Tip) per Tier Loyalitas')
plt.ylabel('Rata-rata Pendapatan ($)')
plt.xlabel('Tier Loyalitas')
plt.show()
```

**Insight untuk Fokus 2: Marketing & ARPU (Segmentasi Tier)**
1. Anomali Sistem Loyalitas: Tier Bronze (pengguna kasual) justru menghasilkan Rata-rata Pendapatan ($24.04) dan Persentase Tip (17.76%) tertinggi dibandingkan semua tier lainnya.

2. Tier Diamond "Pelit": Pelanggan Diamond (tier paling eksklusif) malah memberikan rata-rata pendapatan terendah ($22.56) dan persentase tip yang menurun drastis.

3. Rekomendasi Bisnis:
- Kemungkinan besar, pengguna Diamond mendapatkan terlalu banyak diskon/promo bulanan sehingga mereka membayar lebih murah.
- Tim Marketing harus merombak ulang benefit Tier Diamond. Jangan berikan diskon tarif langsung, melainkan ubah benefit-nya menjadi "Prioritas Pemesanan saat Jam Sibuk" atau "Gratis Upgrade Tipe Mobil", agar pendapatan dasar (ARPU) dari pengguna ini tetap tinggi tanpa mengorbankan loyalitas mereka.

**Fokus 3: Strategi Harga (Surge Multiplier)**

```python
print('=== EDA FOKUS 3: PRICING STRATEGY (SURGE MULTIPLIER) ===\n')

# 1. Membuat kategori/bucket lonjakan harga agar lebih mudah dianalisis
df['surge_bucket'] = pd.cut(
    df['surge_multiplier'], 
    bins=[0.9, 1.1, 1.5, 2.0, 5.0], 
    labels=['1.0x (Normal)', '1.2x - 1.5x', '1.6x - 2.0x', '> 2.0x']
)

# 2. Melihat korelasi Surge vs Status Perjalanan (Apakah harga mahal = banyak dicancel?)
print('Persentase Status Perjalanan berdasarkan Kategori Harga:')
surge_status = df.groupby('surge_bucket', observed=False)['ride_status'].value_counts(normalize=True).unstack() * 100
display(surge_status.round(2))

# 3. Rata-rata User Rating vs Surge Bucket (Apakah harga mahal = rating jelek?)
# Kita abaikan data yang ratingnya -1 (kosong)
rated_rides = df[df['user_rating'] != -1]
print('\nRata-rata Rating Pengguna berdasarkan Kategori Harga:')
surge_rating = rated_rides.groupby('surge_bucket', observed=False)['user_rating'].mean().round(2)
display(surge_rating.to_frame(name='Avg Rating'))

# 4. Visualisasi: Tingkat Pembatalan (Cancellation Rate) vs Lonjakan Harga
# Ambil khusus data yang 'Cancelled'
cancel_rates = surge_status['Cancelled']

plt.figure(figsize=(10, 5))
sns.lineplot(x=cancel_rates.index, y=cancel_rates.values, marker='o', color='red', linewidth=2.5)
plt.title('Tingkat Pembatalan (Cancellation Rate) vs Lonjakan Harga')
plt.ylabel('Persentase Batal (%)')
plt.xlabel('Kategori Lonjakan Harga (Surge Multiplier)')
plt.ylim(0, 20) # Set batas Y agar perubahan terlihat proporsional
plt.show()
```

**Insight untuk Fokus 3: Pricing Strategy (Surge Multiplier)**
1. Harga Mahal Tidak Merusak Permintaan: Saat harga dinaikkan drastis (>2.0x lipat), tingkat penyelesaian pesanan tetap sangat tinggi (78,6%) dan tingkat pembatalan hanya naik sangat sedikit (dari 12,3% menjadi 13,4%).

2. Semakin Mahal, Rating Semakin Bagus: Ajaibnya, perjalanan dengan surge multiplier tertinggi (>2.0x) justru mendapatkan Rating Rata-rata Tertinggi (4.34) dibandingkan perjalanan harga normal (4.22).

3. Rekomendasi Bisnis:
Pelanggan yang memesan di jam sibuk/hujan (saat surge tinggi) sangat menghargai ketersediaan mobil. Mereka tidak peduli dengan harga asalkan mereka bisa pulang/sampai tujuan dengan aman.

Saran: Algoritma Pricing Robotaxi sudah sangat baik. Jangan ragu untuk mempertahankan kebijakan Surge Pricing hingga lebih dari 2.0x di area padat (high-demand), karena terbukti hal tersebut mendongkrak pendapatan secara signifikan tanpa merusak user experience (rating) sama sekali.
